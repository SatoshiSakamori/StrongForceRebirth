# Dimensional comparison only — NOT a Big Rip feasibility proof

This keeps the historical QCD-scale estimate, corrects its interpretation, and does not infer an event horizon, a QCD production rate, or a bounce.

A homogeneous FLRW tidal separation obeys \(\ddot r_{\mathrm{rel}}\sim (\ddot a/a)r=(H^2+\dot H)r\). The former use of \(mrH^2\) omitted \(\dot H\). A proton is a relativistic composite quantum state, so the force comparison below is only an *order-of-magnitude dimensional analogy*.

For \(\sigma\approx1\,\mathrm{GeV/fm}\), \(r=1\,\mathrm{fm}\), \(m=m_p/3\), the ratio \(Hr/c=\sqrt{\sigma r/(mc^2)}\sim1.79\) simply restates comparable chosen energy scales. Cosmological recession rates \(Hr\) are not local relative velocities of freely propagating quarks and crossing \(c\) is not a proof of horizon formation between quarks.

```python
import numpy as np
c = 299792458.0
eV = 1.602176634e-19
sigma = 1.0e9 * eV / 1.0e-15   # N, illustrative 1 GeV/fm
r = 1.0e-15                    # m
m = 1.67262192369e-27 / 3     # kg, illustrative effective constituent mass
H_analogy = np.sqrt(sigma/(m*r))
ratio = H_analogy*r/c
print(f"QCD tension: {sigma:.3e} N")
print(f"Illustrative H scale (NOT derived Big Rip H): {H_analogy:.3e} s^-1")
print(f"H*r/c, a dimensionless ratio only: {ratio:.3f}")
print(f"Equivalent sqrt[(sigma*r)/(m*c^2)]: {np.sqrt((sigma*r)/(m*c*c)):.3f}")
assert np.isclose(ratio, np.sqrt(sigma*r/(m*c*c)))
```
